#### CSV Dataset from Kaggle by Taeef Najib https://www.kaggle.com/datasets/taeefnajib/used-car-price-prediction-dataset

```python
import pandas as pd
import numpy as np

cars = pd.read_csv("used_cars.csv")
cars
```

```python
print(pd.api.types.infer_dtype(cars.price))
```

```python
#the price column is composed of string values, so we need to convert them to float. However, there are dollar and comma signs per element
#removing the dollar and comma signs

cars['price']=cars['price'].str.replace('$', '')
cars['price']=cars['price'].str.replace(',', '')
cars
```

```python
#changing the price column to float:
cars.price = cars.price.astype(float)
print(pd.api.types.infer_dtype(cars.price))
```

```python
cars.describe(include="all")
```

```python
#let's do the same with the mileage column so that the elements can be analyzed numerically
cars['milage']=cars['milage'].str.replace(' mi', '')
cars['milage']=cars['milage'].str.replace(',', '')
cars
```

```python
cars.milage = cars.milage.astype(float)
print(pd.api.types.infer_dtype(cars.milage))
```

```python
cars
```

##### Now the milage and price columns are in float format

```python
cars.describe(include="all")
```

##### Standardizing the price:

```python
cars.insert(12, "standardized_price", 0)
```

```python
# computing and inserting the standardized price values to the new column
cars_price_mean = np.mean(cars.price) #mean of Price
cars_price_std = np.std(cars.price) #std dev of Price

cars.standardized_price = (cars.price - cars_price_mean) / cars_price_std
cars
```

```python
cars.describe(include="all")
```

##### Removing outliers using the 3-sigma method:

```python
cars_milage_mean = np.mean(cars.milage) #mean of milage
cars_milage_std = np.std(cars.milage) #std dev of milage

print ("Cars Milage Mean = " + str(cars_milage_mean))
print ("Cars Milage Stdev = " + str(cars_milage_std))

llc = max(0, cars_milage_mean-3*cars_milage_std)
ulc = cars_milage_mean + 3*cars_milage_std

print("Milage Lower Limit = " + str(llc))
print("Milage Upper Limit = " + str(ulc))
```

```python
cars_cleaned_3sigma = cars[(cars.milage >= llc) & (cars.milage <= ulc)]
cars_cleaned_3sigma    
```

```python
cars_cleaned_3sigma.describe(include="all")
```

```python
#now removing outlier in the price column

print ("Cars Price Mean = " + str(cars_price_mean))
print ("Cars Price Stdev = " + str(cars_price_std))

llc1 = max(0, cars_price_mean-3*cars_price_std)
ulc1 = cars_price_mean + 3*cars_price_std

print("Price Lower Limit = " + str(llc1))
print("Price Upper Limit = " + str(ulc1))
```

```python
cars_cleaned = cars_cleaned_3sigma[(cars_cleaned_3sigma.price >= llc1) & (cars_cleaned_3sigma.price <= ulc1)]
cars_cleaned    
```

```python
cars_cleaned.describe(include="all")
```

##### Here, we can see that there are 2 options for the 'accident' column. This detail is crucial and will affect the price so if there is no input, it's best to remove the records as to not skew the data. We will also do that to missing clean_title inputs.

```python
cars_cleaned=cars_cleaned.dropna(subset=['accident','clean_title'])
cars_cleaned
```

##### For the missing fuel_type data, let's check first:

```python
cars_cleaned.isnull().sum()
```

```python
#checking unique engine types for records with NaN values of fuel_type:

cars_cleaned[cars_cleaned.isnull().any(axis=1)].drop_duplicates(subset='engine', keep='first', inplace=False,ignore_index=True)
```

##### The data above shows that for the 144 records with missing fuel_type values, the engine types of these cars are either all-electric or hybrid vehicles. As such, let us fill the missing values of fuel_type as "Electric/Hybrid"

```python
cars_cleaned['fuel_type']=cars_cleaned['fuel_type'].fillna("Electric/Hybrid")
cars_cleaned
```
