## <span style="color:red">Part1 - 1.</span> Standardize the Price column

```python
import pandas as pd
import numpy as np

data = pd.read_csv("sales.csv")
data
```

```python
#checking the data type of Price series
#could also use:    print(pd.api.types.infer_dtype(data.Price))
data.Price.dtype
```

```python
#Since the series is of string format, it should be converted to float or int
data.Price = data.Price.astype(float)
```

##### As indicated in "ValueError", the data.Price series has the element 13,000 which has a comma so we need to replace it first in order to convert the type of the series into int or float so that the Price data can be manipulated numerically

```python
# searching the index of the element 13,000
print(data.Price[data.Price == "13,000"])
```

```python
#creating a dictionary to replace 13,000 with 13000
price2 = {"13,000":"13000"}

#replacing the 13,000 in the Price series of the data dataframe
data = data.replace({"Price":price2})

data.Price[560]
```

```python
# converting Price series str to float
data.Price = data.Price.astype(float)
print(pd.api.types.infer_dtype(data.Price))
```

```python
#inserting a new column for the standardized Price
data.insert(12, "standardized_price", 0)
```

```python
# computing and inserting the standardized price values to the new column
price_mean = np.mean(data.Price) #mean of Price
price_std = np.std(data.Price) #std dev of Price

data.standardized_price = (data.Price - price_mean) / price_std
```

##  <span style="color:red">Part1 - 2.</span> Identify and remove outliers in the dataset

```python
#Initial characterization of the data
data.describe(include ="all")
```

### Using the 3-sigma method of removing outliers:

```python
m = np.mean(data.Price) #mean of Price
s = np.std(data.Price) #std dev of Price

print ("Price Mean = " + str(m))
print ("Price Stdev = " + str(s))

ll = max(0, m-3*s)
ul = m + 3*s

print("Price Lower Limit = " + str(ll))
print("Price Upper Limit = " + str(ul))
```

# ------------------------------------------------------------------------------
##### Before cleaning the data using the 3-sigma method, I would like to see information on the Prices first

```python
#Inserting a Count column to get the number of Prices in the dataframe, but used a copy of the data
data_temp = data.copy()
data_temp.insert(13, "Count", 0)

#confirming if original data weren't changed (should have no new column named Count)
data
```

```python
#Assigning values to the Count column - number of times a specific Price element appeared in the Price column

for i in range (len(data_temp['Price'])):
    data_temp.Count[i] = np.count_nonzero(data.Price==data.Price[i])
    
data_temp
```

```python
#Removing duplicate Prices to the data_temp to show the unique prices for the products
#Named another dataframe to hold those unique values

Unique_Prices = data_temp.drop_duplicates(subset='Price', keep='first', inplace=False, ignore_index=False)
print(Unique_Prices[['Product','Price','Count']])

print("\nNumber of unique prices: " + str(np.count_nonzero(Unique_Prices.Price)))
```

##### With this Unique_Prices dataframe shown, an initial inspection of the Products and Prices would lead to the conclusion that there should only be 3 prices in the data - one for each Product (Product1, Product2, Product3). The other prices are only one-time input errors for the Product1 price.

##### As such, I believe that the best method to clean the data and remove outliers is by  <span style="color:blue">REMOVING INPUT INACCURACIES</span> in the Product1 Price data. This is because when we use the 3-sigma method, since the price's upper limit that was computed earlier was 5280.88 or below the Product3 price, the Product3 will not be included anymore in the data analysis. Product3 will be removed and treated as an outlier even though it is essential in the data.

```python
#Removing inaccuracies in the Price:

data_cleaned_inaccuracies = data[(data.Price == 1200) | (data.Price == 3600) | (data.Price == 7500)]

data_cleaned_inaccuracies.describe(include="all")
```

```python
data_cleaned_inaccuracies
```

##### The above shows the data for the sales of the 3 products with their corresponding true prices (rows with inaccurate prices were removed)
### --------------------------------------------------------------------------------

### For reference purposes, I will still clean the data using the 3-sigma approach and the IQR approach

#### Continuing with the 3-sigma approach of removing outliers:

```python
print ("Price Mean = " + str(m))
print ("Price Stdev = " + str(s))
print("Price Lower Limit = " + str(ll))
print("Price Upper Limit = " + str(ul))

data_cleaned_3sigma = data[(data.Price >= ll) & (data.Price <= ul)]
data_cleaned_3sigma                                                 
```

```python
data_cleaned_3sigma.describe(include="all")
```

##### As seen above, using the 3-sigma limit approach of removing outliers in these data will result in removing a product with the correct pricing (product3). Upon cleaning the data by subjecting them to the upper and lower limits, the minimum Price value is still 20 (which is obviously an error because as stated in the Count column of the Unique_Prices dataframe, Product1's Price should be 1200), and the max value is now 3600 (price of Product2), removing the rows for the Product3 (priced at 7500). The spread of the Price data may be lower (low std), but this results in an incomplete data which could negatively affect the analysis.

### Now let's try the Inter-Quartile Range (IQR) method:

IQR = Q75 - Q25

```python
IQR = np.percentile(data.Price,75) - np.percentile(data.Price,25)

print("75th percentile of Price data column = " + str(np.percentile(data.Price,75)))
print("25th percentile of Price data column = " + str(np.percentile(data.Price,25)))
print("\n\b IQR = " + str(IQR))

ul2 = np.percentile(data.Price,75) + 1.5*IQR
ll2 = max(0, np.percentile(data.Price,25)-1.5*IQR)

print("\nUpper limit 2 = " + str(ul2))
print("Lower limit 2 = " + str(ll2))

data_cleaned_IQR = data[(data.Price >= ll2) & (data.Price <= ul2)]

print('\033[1m' + "\nCleaned data using the IQR method: ")
data_cleaned_IQR
```

##### This shows that for this dataframe, the IQR approach in removing outliers SHOULD NOT be used. It will only result in removing all of the data since the IQR computed is nan (not a number) which results in considering all of the prices as outliers when they are subjected to the limits.

--------------------------------
#### We can see that there are 3 columns that have missing values. It's good that there are no missing values in the Buy_Car column (if there are, I think it's best to delete those rows with missing Buy_Car input because it's the target variable).

#### There are 7 columns, and I think it's best for a specific row or record if 5 of those have no missing values so that there will be enough information for the data. As such, those with 3 or more missing inputs will be deleted

```python
# deleting all rows with less than 5 non-NaN values
part2 = part2.dropna(thresh = 5)
print("Number of rows: " + str(part2['ID'].count()))
part2
```

#### Only 1 record was deleted.
-----------------------------------------
#### For the 'Sex' column, I will FILL the missing values with "No Data" because the respondents might have purposefully skipped that. They could have not included one because they could have a different sexual orientation that was not specified in the options. 

#### For the 'Children' column, I will KEEP the missing values since the respondents who missed that could have a purpose for not disclosing the number of their children, like for privacy or safety purposes. Filling those missing values with a "No Data" string will mix a string data type to the numerical column, so it should be avoided if we want to analyze the Children column in the future using functions specialized for numbers (like max, min, mean). 

```python
#fill NaN values of Sex column
part2['Sex'] = part2['Sex'].fillna("No Data") 
part2
```

--------------------------------
#### For the 'Income' column, I will handle the missing values using two ways: fill or keep. 

First, if the respondent is <span style="color:blue">Single</span> and will buy a car <span style="color:blue">(Buy_Car = Yes)</span>, I will use the <span style="color:blue">MEAN</span>  to fill up the missing income cell (the distribution of incomes is normal so a mean is preferred) because I think there is a high chance for the respondent to be earning if he or she wants to buy a car even without a partner to rely on for the cost and with the given age range of the respondents (25-30 yrs old is the age range where people have usually stopped asking for gifts or financial support from their parents). 

Second, I will <span style="color:blue">KEEP</span>  the missing income value if the respondent is <span style="color:blue">Married</span>  since this could be meaningful. He or she could be relying on the income of his or her partner (e.g. she's a housewife, or unemployed). I will also keep the missing value if he/she is single and will NOT buy a car because this could indicate that he/she is unemployed.

```python
#Changing the wrong column name: "Employed" should be "Marital Status" based on the column content
part2.rename(columns={'Employed' : 'Marital_Status'}, inplace=True)
part2
```

```python
# querying (reference only to show the data that satisfy the conditions)
part2_queried = part2[(part2.Marital_Status=='Single') & (part2.Buy_Car=='Yes')]
part2_queried
```

```python
#getting the mean of incomes
income_mean = part2['Income'].mean() #income_mean = part2.Income.mean()
print("Income_mean = " + str(round(income_mean,2)))
```

```python
#resetting the index of the dataframe for easier manipulation
part2 = part2.reset_index(drop=True)
#reset_index func adds another column for old indices, default drop=False
#drop=True means dropping the newly generated columns for old indices
#ignore_index=True parameter is not working for dropna function

part2
```

##### Notice the indices above, they are now arranged or ordered properly so that we will not have a problem in the loop below

```python
#filling in missing values for the queried data:
for x in range (len(part2['Income'])):
    if ((part2.Marital_Status[x]=='Single') and (part2.Buy_Car[x]=='Yes')):
        if part2.Income[x] > 0:
            part2.Income[x] = part2.Income[x]
        else:
            part2.Income[x]=income_mean
            
part2    
```

##### ID 18 now has its income value filled with the Mean. It's the only row that satisfied the conditions (Income  = NaN, Buy_car = Yes, Marital_Status = Single).

##### The other missing income value was kept as is, for reasons stated above.
